# Lesson 1 — What a language model is

A language model has one job: given the words so far, put a probability on every word that could come next,
P(next word | words so far). Autocomplete shows the top few of those; text generation keeps sampling from them.

## Why we only look back a few words

The chain rule splits a sentence's probability into one next-word step per word. But conditioning on the *whole*
history is hopeless — most long prefixes never appear in any training text. The Markov shortcut says only the last
n-1 words matter: a **bigram** looks at 1 previous word, a **trigram** at 2. After that, it's just counting.

## A hand-made probability table

Before learning anything from data, here is what a model *is*: a lookup table from a context word to a set of
next-word probabilities. These numbers are invented.

```python
# A language model is just a probability table over next words. Here's a hand-made one:
next_word = {
    "the": {"morning": 0.30, "train": 0.20, "park": 0.15, "city": 0.10, "weekend": 0.08},
    "good": {"book": 0.5, "coffee": 0.5},
}
```

`suggest` sorts the candidates for a context by probability and keeps the top `k` — the suggestion bar on a
phone keyboard.

```python
def suggest(context, table, k=3):
    ranked = sorted(table.get(context, {}).items(), key=lambda kv: -kv[1])
    return ranked[:k]
```

```python
print("after 'the' :", suggest("the", next_word))
print("after 'good':", suggest("good", next_word))
```

The ranking works, but I made up every number. The rest of the course learns these numbers from real text
by counting — starting in Lesson 2.

## Test yourself

**1. What does a language model actually compute?**
A probability for each possible next word given the words so far, P(next | context). Autocomplete shows the top
few; generation samples from it again and again.

**2. Why make the Markov assumption instead of using the full history?**
Almost every long prefix is unique, so you could never count how often a word follows it. Keeping only the last
n-1 words gives contexts that repeat often enough to count.

**3. What context does a trigram model use for P(wᵢ | …)?**
The two previous words, wᵢ₋₂ and wᵢ₋₁. A bigram uses only wᵢ₋₁.
